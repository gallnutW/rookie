## 1. `try...except...finally...END`
> `except` 会捕获 `try` 块里面出现的异常
>
> `finally` 块必定执行
>
> 常见的错误类型和继承关系 `https://docs.python.org/3/library/exceptions.html#exception-hierarchy`
>
> 父类异常不能在子类异常之前，否则永远不会触发子类异常的捕获（在父类异常的部分就被捕获了）
>
> 调用栈——如果错误没有被捕获，它就会一直往上抛，最后被Python解释器捕获

```python
def try_example(x):
    try:
        print('try...')
        r = 10 / x
        print('result:', r)
    except TypeError as e:
        print("(Exception)TypeError: ",e)
    except ValueError as e:
        print('(Exception)ValueError:', e)
    except ZeroDivisionError as e:
        print('(Exception)ZeroDivisionError:', e)
    finally:
        print('finally...')
    print('END')
    print('-'*75)

try_example(0)
try_example(2)
try_example('0')
```

## 2. 记录错误—— `logging` 模块
> `logging.exception()` 必须且只能在捕获异常的 `except` 块中使用
>
>

```python
import logging
def try_example(x):
    try:
        print('try...')
        r = 10 / x
        print('result:', r)
    except TypeError as e:
        print("(Exception)TypeError: ",e)
    except ValueError as e:
        print('(Exception)ValueError:', e)
    except ZeroDivisionError as e:
        print('(Exception)ZeroDivisionError:', e)
        logging.exception(e)
    finally:
        print('finally...')
    print('END')
    print('-'*75)

try_example(0)
print("Continue...")
```

 ## 3. 抛出错误

```python
class MyValueError(ValueError):
    pass
def calculate_example(x):
    if x == 4:
        raise MyValueError("I hate number 4")
    r = 10 / x
    return r
```

```python
calculate_example(4)
```

- 当 raise 语句后面不接任何参数时，它的含义是：“把当前 except 捕获到的这个异常，原封不动地重新向外层调用栈抛出。”

```python
def try_example(x):
    try:
        print('result:', calculate_example(x))
    except MyValueError as e:
        print("(Exception)MyValueError:", e)
        raise
def call_try_example(x):
    try:
        try_example(x)
    except MyValueError as e:
        print("try_example() throws MyValueError to me(the caller)")
        logging.exception(e)
    finally:
        print('finally...')
```

```python
try_example(2)
try_example(4)
```

```python
call_try_example(2)
print("-"*75)
call_try_example(4)
```

## 4. `logging` 的使用
> logging不会抛出错误，而且可以输出到文件
>
> logging.basicConfig()设置基本参数
>
> PyCharm 的 Jupyter(底层是 ipykernel/IPython)在启动 kernel 时,通常已经提前给 root logger 挂载了一个 handler
>
> 导致logging.basicConfig被忽略掉了，所以需要加入force=True参数（root logger 是这个进程内的全局单例，重启进程又会恢复（已验证））
>

```python
import logging

# logging.basicConfig(level=logging.INFO)
logging.basicConfig(level=logging.INFO, force=True)
s = '0'
n = int(s)
logging.info(f"n = {n}")
print(10 / n)
```
