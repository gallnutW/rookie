# 面向对象
> 封装，继承，多态
>
> 在Python中，所有数据类型都可以视为对象，
>
> 当然也可以自定义对象。自定义的对象数据类型就是面向对象中的类（Class）的概念

## 类和实例演示

```python
from pyclbr import Class


class Student(object):
    def __init__(self, name="GallnutW", score="100"):
        self.name = name
        self.score = score
        print(f"正在初始化实例，执行__init__,\n该实例:{self.name}, score={self.score}")
    def print_score(self):# 封装
        print('%s: %s' % (self.name, self.score))
student=Student()
student.print_score()
```

## 访问限制
> 在Python中，实例的变量名如果以__开头，就变成了一个私有变量（private）
>
> 通过get(self)来访问私有变量，set(self)修改私有变量（可以执行参数检查）
>
> 事实上__attr可以内外部访问，不过前面要加上类名前缀,比如声明类A，里面有私有变量__B，可以用 _A__B访问
>
> __xxx__属于特殊变量是可以直接访问的

```python
class Student(object):
    def __init__(self, name="GallnutW", score="100", __weight=80):
        self.name = name
        self.score = score
        self.__weight = __weight
        print(f"正在初始化实例，执行__init__,\n该实例:{self.name}, score={self.score}, weight={self.__weight}")
    def print_score(self):# 封装
        print('%s: %s' % (self.name, self.score))
    def get_weight(self):
        return self.__weight
    def set_weight(self, weight):
        if 0 <= weight <= 100:
            self.__weight = weight
        else:
            raise ValueError("分数只能在[0, 100]中")
student = Student()
# print(student.__weight)会执行失败
print(student._Student__weight)
print(student.get_weight())
# set_weight(120)
student.set_weight(100)
print(student.get_weight())
```

## 继承和多态
> 子类Subclass ***继承*** 父类Baseclasss/Superclass
>
> 多态——子类可以覆盖父类的方法
>
> 鸭子类型——对象的有效语义不是由它继承的特定类或实现的接口决定的，而是由它当前具备的方法和属性集合决定的

```python
class Animal():
    def __init__(self, __name, __food):
        self.__name = __name
        self.__food = __food
        self.size = 9
    def get_name(self):
        return self.__name
    def get_food(self):
        return self.__food
    def set_name(self, name):
        self.__name = name
    def set_food(self, food):
        self.__food = food
    def say_hi(self):
        print(f"{self.__name} is saying hi")
class Dog(Animal):
    def say_hi(self):
        print(f"{self.get_name()} is WANG WANG WANG WANG")

dog=Dog("a", "bone")
dog.say_hi()
print(dog.get_food())
```

```python
a = Animal("a", "bone")
d = Dog("b", "bone")
print(isinstance(a, Animal), isinstance(d, Dog), isinstance(d, Animal))

print(type(d),type(d.say_hi()))
```

```python
class Duck:
    def quack(self):
        print("嘎嘎嘎！")

class Person:
    def quack(self):
        print("我在模仿鸭子嘎嘎叫！")

class Car:
    def honk(self):
        print("滴滴滴！")

# 这个函数不关心传入的 thing 是什么类型（类），只关心它有没有 quack() 方法
def make_it_sound(thing):
    thing.quack()

duck = Duck()
person = Person()
car = Car()

make_it_sound(duck)   # 正常执行：输出 "嘎嘎嘎！"
make_it_sound(person) # 正常执行：输出 "我在模仿鸭子嘎嘎叫！" (无需继承鸭子类)
# make_it_sound(car)  # 报错：AttributeError，因为 Car 对象没有 quack() 方法
```

## 类属性与实例属性
- 给实例绑定属性的方法是通过实例变量，或者通过self变量
- 直接在class里面定义的属性就是类属性
- 可以给实例也绑定属性（或者是方法）
- 实例属性优先级比类属性高

```python
class Student(object):
    name = 'Student'

s = Student() # 创建实例s
print(s.name) # 打印name属性，因为实例并没有name属性，所以会继续查找class的name属性

print(Student.name) # 打印类的name属性

s.name = 'Michael' # 给实例绑定name属性
print(s.name) # 由于实例属性优先级比类属性高，因此，它会屏蔽掉类的name属性

print(Student.name) # 但是类属性并未消失，用Student.name仍然可以访问

del s.name # 如果删除实例的name属性
print(s.name) # 再次调用s.name，由于实例的name属性没有找到，类的name属性就显示出来了

def hello():
    return "Hello World!"
s.hello = hello()
print(s.hello)
```
