# 如何 `import module`

## 关于 Package
- 1. 导入一个包的时候，Python 会根据 sys.path 中的目录来寻找这个包中包含的子目录
- 2. 只有包含 `__init__.py` 的文件夹才能被识别为包（3.3+之后，不加入该文件也可以被识别为包，但是最好还是加上）

> `sys.path` 包含
>
> 1. 当前执行脚本所在目录
> 2. PYTHONPATH 环境变量指定的目录
> 3. 标准库（OS，json等标准库）目录

```python
import sys

from AboutPython.aboutModule.my_module.multiply_tool import multiply

print(sys.path)
print("-" * 75)
print("-" * 75)
print(sys.modules)
```

### 如何导包
> 这里主要是讲本地项目里面的模块导入
> 1. 在执行 `import` 的时候，python 都是从 `sys.path` 里面找的
> 2. 对于本地项目而言，查找的路径就是从项目根路径开始的，也就是 `sys.path` 里的第一个路径
> 3. 项目根目录下的模块能够直接被导入
> 4. 如果不在根目录，而是在子目录，则用点号 `.` 把包结构（文件结构展示出来）
> 5. 涉及到多层结构的时候，可以使用 `import a.b.c as d` 来简化后续的使用

```python
# 通过这种重命名方式简化使用，可以看到下面的打印结果是一样的

import my_module
import my_module.my_module_func as my_func
print(my_func)
print(my_module.my_module_func)
my_module.my_module_func.greeting()
my_func.greeting()
```

> 6. python在导入包的时候默认不导入子模块，所以可以看到下面这个子模块还需要重新导入

```python
import my_module.my_sub_module_A.sub_func
print(my_module.my_sub_module_A)
my_module.my_sub_module_A.sub_func.goodbye()
```

> 7. 如果不想重新导入子模块，可以在父模块的 `__init__.py` 里面加入导入语句
> - 实际上，在执行import的时候，被导入的包的__init__.py会被完整执行一遍

```python
import my_module
print(my_module.my_sub_module_B)
print(my_module.my_sub_module_B.sub_func_B)
my_module.my_sub_module_B.sub_func_B.goodbye()
```

> 8. 项目里面多个不同层次的模块相互导入的方法（相对，间接）
> > 见 `my_sub_module_C` 里面的示例
> >
> > 相对导入太容易出问题了，经过尝试还是尽可能选择绝对导入吧
