### 这里记载一些关于git的使用

- git config --global --list查看本地git的全局配置
- git config --global user.name "Gallnut" 设置用户名
- git config --global user.email "gallnut9904@gmail.com" 设置用户邮箱(这也是我github的账号邮箱)

- git remote add origin https://...../*.git  将origin设为github某一远程仓库的别名
- git remote -v 查看当前仓库关联了哪些远程仓库的命令
- git status 可以查看本地的分支名，git branch -M main可以将本地分支名强制改为main
- 为什么要改？因为老版git的默认分支是master，而github是main，直接推送上去，会在github上创建一个新的master分支
- git remote -u origin <本地分支名>      -u简历本地分支与远程分支（origin）的追踪关系，origin是推送的目标

第二台电脑git push时会验证身份
