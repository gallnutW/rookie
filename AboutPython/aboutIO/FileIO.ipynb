# 文件读写

> - `'r'` open for reading (default)
>
> - `'w'` open for writing, truncating the file first
>
> - `'x'` open for exclusive creation, failing if the file already exists
>
> - `'a'` open for writing, appending to the end of file if it exists
>
> - `'b'` binary mode
>
> - `'t'` text mode (default)
>
> - `'+'` open for updating (reading and writing)

## 1. 读文件
> `f = open(file_addr, open_mode)`
>
> `open_mode` 有 `'r'`
<br>
>
>  `with open(file_addr_01, "r") as f:`  with语句来自动调用close()

```python
root_dir = "files"
file_addr_01 = root_dir + "/file01.txt"

try:
    with open(file_addr_01, "r") as f:
        content = f.read()
        print(type(content))
        print(content)
except FileNotFoundError as e:
    print(e)
```

- `read()` 一次性读取所有文件内容，指针会直接指向文件末尾，所以，连续执行两次read，第二次read是空的
- `readlines()` 一次读取一行

```python
with open(file_addr_01, "r") as f:
    contens = f.readlines()
    print(type(contens))
    for line in contens:
        print(line.strip())
```

- file-like Object
- 除了file外，还可以是内存的字节流，网络流，自定义流等等。file-like Object不要求从特定类继承，只要写个read()方法就行。

- 二进制文件

```python
picture_01_addr = root_dir + "/picture01.png"
with open(picture_01_addr, "rb") as f:
    content = f.read()
    print(type(content))
    print(content)
```

## 2.写文件
> 只有调用close()方法时，操作系统才保证把没有写入的数据全部写入磁盘
>

```python
with open(file_addr_01, 'r+') as f :
    contens_origin = f.readlines()
    f.write('\n')
    for line in contens_origin:
        print(line.strip())
        f.writelines(f"copy_version:{line}")
```

# StringIO 和 BytesIO
> 在内存中读写

### StringIO
> 操作str

```python
from io import StringIO
f = StringIO()
f.write("Hello! ")
f.write("Rookie! ")
print(type(f.getvalue()))
value = f.getvalue()
print(value)
```

### BytesIO
> 操作二进制数据

```python
from io import BytesIO
f = BytesIO()
input_str = '中文'
f.write(input_str.encode('utf-8'))

output = f.getvalue()
print(type(output))
print(output)
```

# 操作文件和目录
> 用到 `os` 模块

```python
import os

# print(os.uname())   # 详细系统信息，Windows不提供该功能
print(os.name)      # 打印当前操作系统
print(os.getcwd())  # 打印工作目录
print(os.listdir()) # 取当前工作目录的内容
print(os.environ)   # 操作系统中定义的环境变量

import platform

print(platform.system())   # 输出操作系统名称，如: 'Windows'
print(platform.release())  # 输出版本号，如: '10' 或 '11'
print(platform.version())  # 输出详细构建版本号
print(platform.machine())  # 输出架构，如: 'AMD64'
```

### 文件和目录操作函数
> 查看当前目录 `os.path.abspath('.')`
>
> 目录拼接 `os.path.join(root_dir, file_dir)`——用函数拼接可以避免路径拼接符出现问题
>
> 创建目录 `os.mkdir(dir_path)`
>
> 删掉目录 `os.rmdir(dir_path)`
>
> 路径分割 `os.path.split()` 这样可以把一个路径拆分为两部分，后一部分总是最后级别的目录或文件名
>
> `os.path.splitext()` 可以获得文件扩展名
>
> 文件重命名 `os.rename` (old_file_addr. new_name)
>
> 删掉文件 `os.remove'
>

```python
root_dir = os.path.abspath('.')
new_dir_name = "new_dir"

target_dir = os.path.join(root_dir, new_dir_name)
print(target_dir)

print(os.getcwd())
print(os.listdir())

os.mkdir(target_dir)
print(os.listdir())
print(os.path.split(target_dir))
```

```python
new_file_addr = os.path.join(target_dir, "new_file.txt")
print(new_file_addr)
with open(new_file_addr, "w") as f:
    f.write("Hello!")
```

```python
new_file_name = os.path.join(target_dir, "new_file_rename.txt")
os.rename(new_file_addr, new_file_name)
```

```python
os.remove(new_file_name)
os.rmdir(target_dir)
print(os.listdir())
```

# 序列化
> 把变量从内存中变成可存储或传输的过程称之为序列化
>
> 变量内容从序列化的对象重新读到内存里称之为反序列化
>

> `pickle.dumps()` 方法把任意对象序列化成一个 `bytes`

```python
import pickle
d = dict(name="bob", age=22)

d_serialized = pickle.dumps(d)
print(type(d_serialized))
print(d_serialized)
```

```python
target_dir = "files"
dump_file = "dump_file.txt"

target_file_addr = os.path.join(target_dir, dump_file)
with open(target_file_addr, "wb") as f:
    f.write(d_serialized)
```

```python
with open(target_file_addr, "rb") as f:
    print(f.read())
    f.seek(0)
    original = pickle.load(f)
    print(type(original))
    print(original)
```

### JSON

- ___jsonToPython___
> {}            dict
>
> []            list
>
> "string"      str
>
> 1234.56       int/float
>
> true/false    True/False
>
> null          None

```python
import json
print(d)
json_file = json.dumps(d)
print(type(json_file))
print(json_file)
```

```python
d_back = json.loads(json_file)
print(type(d_back))
print(d_back)
```

- 序列化其他格式(通过dumps里面的参数)
> 可选参数`default`就是把任意一个对象变成一个可序列为JSON的对象（通过写好的转换函数）

```python
class Student(object):
    def __init__(self, name, age, score):
        self.name = name
        self.age = age
        self.score = score
# Student-->dict
def student2dict(std):
    return {
        'name': std.name,
        'age': std.age,
        'score': std.score
    }
# json.dumps(), dict-->json
# json.loads(), json-->dict
# dict-->Student
def dict2student(d):
    return Student(d['name'], d['age'], d['score'])

s = Student('Bob', 20, 88)
student_serialized=json.dumps(s, default=student2dict)
print(type(student_serialized))
print(student_serialized)

student_back = json.loads(student_serialized)
print(type(student_back))
print(student_back)

student_original = dict2student(student_back)
print(type(student_original))
student_original
```
