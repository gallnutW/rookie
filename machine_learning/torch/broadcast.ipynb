# 张量广播机制

- ___张量的广播原则___
> 1. 每个张量至少有一个维度（**空张量不能广播**）。
> 2. 比较两个张量的形状时，必须从尾部（**_从右向左_**）开始逐个比较维度。
> 3. 从右向左比较时，两个维度的大小必须满足以下三种情况之一：
> > 1. 维度大小相等。
> > 2. 其中一个维度大小为 1（大小为 1 的维度会被复制扩展以匹配另一个张量的维度大小）。
> > 3. 其中一个张量没有该维度（缺失的维度会被视为 1，然后再进行扩展）。

```python
import torch
```

- 张量与标量

```python
a = torch.ones(2,3)
print(a)
b = 1
print(a + b)
```

- 维度较少的张量向多维张量广播（缺失维度补齐）

```python
a = torch.ones(4, 3, 2)  # shape: (4, 3, 2) 这里可以试一试改成 4 * 4 * 2 和 4 * 1 * 2
b = torch.ones(3, 2)     # shape:    (3, 2)

# 从右向左比：
# 尾部第1维: 2 和 2 (相等，通过)
# 尾部第2维: 3 和 3 (相等，通过)
# 尾部第3维: 4 和 缺失 (缺失被当作 1，然后扩展为 4，通过)
c = a + b                # c shape: (4, 3, 2)
print(c.shape)
print(c)
```

```python
a = torch.ones(3, 1)  # shape: (3, 1)
b = torch.ones(1, 4)  # shape: (1, 4)

# 比较规则：
# 尾部第1维: 1 和 4 (a的1扩展为4)
# 尾部第2维: 3 和 1 (b的1扩展为3)
c = a + b             # c shape: (3, 4)

```

```python
# 上述这种方式的一个更常见情况是这样的
a = torch.ones(5, 4) # 代表5个样本，每个样本4维
b = torch.ones(2, 4) # 代表2个样本，每个样本4维

# 现在要让a中每个样本和b中每个样本两两相加，得到十个结果，就是这样
c = a[:, None, :] + b[None, :, :]
# 等价写法
d = a.unsqueeze(1) + b.unsqueeze(0)
print(c.shape)
print(d.shape)
```

```python
print(c.flatten(0, 1).shape) # 将第0维到第1维拉平
print(d.flatten(0, 1).shape)
```
