# 如何 `import module`

## 关于 Package
- 1. 导入一个包的时候，Python 会根据 sys.path 中的目录来寻找这个包中包含的子目录
- 2. 只有包含 `__init__.py` 的文件夹才能被识别为包

> `sys.path` 包含
>
> 1. 当前执行脚本所在目录
> 2. PYTHONPATH 环境变量指定的目录
> 3. 标准库（OS，json等标准库）目录

```python
import sys
print(sys.path)
```

```python
print(sys.modules)
```

### 如何导包
> 这里主要是讲本地项目里面的模块导入
> 1. 在执行 `import` 的时候，python 都是从 `sys.path` 里面找的
> 2. 对于本地项目而言，查找的路径就是从项目根路径开始的，也就是 `sys.path` 里的第一个路径
> 3. 项目根目录下的模块能够直接被导入
> 4. 如果不在根目录，而是在子目录

```python
import my_module.my_module_func
print(my_module)
my_module.my_module_func.greeting()
```
