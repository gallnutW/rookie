## 列表生成式
- ... for () in () ...
- for前面加表达式，后面加if筛选条件

```python
l1 = list(range(1, 11))
print(l1)

l2 = list(x * x for x in range(1, 11))
print(l2)

l3 = list(x * x for x in range(1, 11) if x % 2 == 0)
print(l3)

l4 = list(m +'_'+ n for m in 'ABC' for n in 'XYZ')
print(l4)

# 从[1,10]里面选择3的倍数，如果该数是偶数，那么取原数，否则取相反数
l5 = list(x if x % 2 == 0 else -x for x in range(1, 11) if x % 3 == 0)
print(l5)
```

## 生成器
- 法一：将列表生成式的[]改成()
- 加入yield得到generator函数

```python
L = [x * x for x in range(10)]
print(L)

g = (x * x for x in range(3))
print(g)
print(isinstance(g, Iterable))
print('---')
for i in g:
    print(i)
print('----')
for i in range(3) :
    print(next(g))
```

- 普通函数——顺序执行，遇到return语句或者最后一行函数语句就返回
- generator函数——在每次调用next()的时候执行，遇到yield语句返回，再次执行时从上次返回的yield语句处继续执行。
- 多次调用generator函数会创建多个相互独立的generator
- 捕获StopIteration错误，返回值包含在StopIteration的value中

```python
def fib(max):
    n, a, b = 0, 0, 1
    while n < max:
        yield b
        a, b = b, a + b
        n = n + 1
    return 'done'

f = fib(5)
print(f)
print('---' * 10)
for n in f:
    print(n)

# 创建了多个独立的generator对象
print('---' * 10)
for i in range(5):
    print(next(fib(5)))
```

```python
g = fib(6)
while True:
    try:
       x = next(g)
       print('g:', x)
    except StopIteration as e:
       print('Generator return value:', e.value)
       break
```
