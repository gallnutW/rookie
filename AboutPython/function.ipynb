- 参数检查

```python
# 参数检查
def my_abs(x):
    if not isinstance(x, (int, float)):
        raise TypeError('bad operand type')
    if x >= 0:
        return x
    else:
        return -x
my_abs('A')
```

- 函数可以返回多个值
- （多个值封装在tuple里，所以实际上还是返回的一个值）

```python
# return multiple results(Actually a tuple with multiple results)
import math

# argument "angle" is a default argument
# default argument musts be place behind necessary arguments

def move(x, y, step, angle=0.0):
    nx = x + step * math.cos(angle)
    ny = y + step * math.sin(angle)
    return nx, ny

print(type(move(0,0,1,1)))
move(0,0,1.0, math.pi/2)
```

- 默认参数的使用
- 默认参数需要指向不变对象（str，int，tuple， None等）

```python
# default arguments must point to unchangeable objects!!!
print('-' * 15, "默认参数指向了可变对象", "-" * 15)
def add_end(L=[]):
    L.append('END')
    return L
print(add_end())
print(add_end())

print('-' * 15, "默认参数指向了不变对象", "-" * 15)
def add_end(L=None):
    if L is None:
        L = []
    L.append('END')
    return L
print(add_end())
print(add_end())
```

- 可变参数
- 即传入的参数个数是可以改变的
- 可变参数在函数调用时自动组装为一个tuple
- Python允许你在list或tuple前面加一个*号，把list或tuple的元素变成可变参数传进去

```python
def calc(*numbers):
    sum = 0
    for n in numbers:
        sum = sum + n * n
    return sum
print(calc(1, 3, 5, 7))

nums = [1, 3, 5, 7]
print(calc(*nums))
```

- 关键字参数
- 允许你传入0个或任意个 ***含参数名*** 的参数，这些关键字参数在函数内部自动组装为一个dict

```python
def person(name, age, **kw):
    print('name:', name, '\nage:', age, '\nother information about this man:', kw)
print(person('Michael', 30))
print(person('Adam', 45, gender='M', job='Engineer'))
```

```python
def person(name, age, **kw):
    if 'city' in kw:
        # 有city参数
        print(f"\nHe lives in {kw.get('city')}")
        pass
    if 'job' in kw:
        # 有job参数
        print(f"\nHe has A JOB,it is {kw.get('job')}")
        pass
    print('name:', name, '\nage:', age, '\nother:', kw)
print(person('Michael', 30))
print(person('Adam', 45, gender='M', job='Engineer'))
```

- 命名关键字参数
- 用于限制关键字参数的名字
- 加入分隔符 * ，* 后的参数被视为命名关键字参数
- 如果函数定义中已经有了一个可变参数，后面跟着的命名关键字参数就不再需要一个特殊分隔符

```python
def person(name, age, *, city, job):
    print(name, age, city, job)
print(person('Jack', 24, city='Beijing', job='Engineer'))

#如果函数定义中已经有了一个可变参数，后面跟着的命名关键字参数就不再需要一个特殊分隔符
def person(name, age, *args, city, job):
    print(name, age, args, city, job)
print(person('Michael', 24, city='Beijing', job='Engineer'))
# TypeError: person() missing 2 required keyword-only arguments: 'city' and 'job'
# print(person('Michael', 24,'Beijing','Engineer'))

# 当命名关键字参数有default value时，不需要传入参数
def person(name, age, *args, city = 'Beijing', job='Engineer'):
    print(name, age, args, city, job)
print(person('Michael', 24,'Beijing','Engineer'))
```
