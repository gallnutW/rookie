# `with ... as ...` 与 `async with ... as ...`

`with as`
> 1. 接收一个类对象
> 2. 该类对象必须实现了 `__enter__` 和 `__exit__`
> 3. `__enter__` 在进入的时候调用，返回结果赋值给 `as` 后的那个对象
> 4. `__exit__` 在退出的时候执行

```python
from dataclasses import dataclass, field
@dataclass
class MyResource:
    name : str
    def __enter__(self):
        """进入with块时调用，as后的变量就是这个方法的返回值"""
        print("现在正在执行__enter__, 获取资源")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        """退出with块时调用，即使发生异常也会执行"""
        print("释放资源")
        # 返回True表示抑制异常，返回False/None表示继续抛出异常
        return False
    def using(self):
        print(f"{self.name} 正在使用资源")
# 使用
with MyResource("Alice") as res:
    print(type(res))
    print(res)

    print("使用资源")
    res.using()
```

`async with as`
> 1. 异步代码中管理异步资源的获取和释放
> 2. 只能写在 `async def` 声明的函数里面（在 jupyter 里面写的时候不用，因为自带的有一个全局事件循环）
> 3. 该类对象必须实现了 `__aenter__` 和 `__aexit__`
> 4. 然后用`asyncio.run()`来执行前面声明的异步函数
> 5. `asyncio.run()`是启动一个事件循环
> 6. `await` 是在事件循环内等待异步操作

```python
import asyncio
class AsyncResource:
    async def __aenter__(self):
        """进入async with块时调用"""
        print("异步获取资源")
        await asyncio.sleep(1)  # 模拟异步操作
        return self

    async def __aexit__(self, exc_type, exc_val, exc_tb):
        """退出async with块时调用"""
        print("异步释放资源")
        await asyncio.sleep(0.5)  # 模拟异步清理
        return False

# 使用
async def main():
    async with AsyncResource() as res:
        print("使用异步资源")

# asyncio.run(main()) jupyter里用不了这个
await main()
```
